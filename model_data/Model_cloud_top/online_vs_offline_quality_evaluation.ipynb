```python
import cartopy.crs as ccrs


p = offline_ct.ctt_filtered.isel(time=1).plot( subplot_kws=dict(projection=ccrs.Orthographic(0, 0), facecolor="gray"), transform=ccrs.PlateCarree(), vmin=250)


p.axes.coastlines()
```
